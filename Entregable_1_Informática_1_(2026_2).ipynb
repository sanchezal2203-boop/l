### 2. Estructura condicional con banderas (20%)

Un sistema de monitoreo básico recibe tres signos vitales de un paciente y debe: (1) clasificar su frecuencia cardíaca, y (2) determinar si el paciente requiere atención inmediata, usando una bandera que combine varios factores de riesgo.

El programa debe:

- Leer la frecuencia cardíaca fc (entero, en latidos por minuto).
- Leer la temperatura corporal temperatura (float, en °C).
- Leer la saturación de oxígeno saturacion (entero, en %).

Clasificar fc en una variable clasificacion:
- `fc < 60 → "bradicardia"`
- `60 <= fc <= 100 → "normal"`
- `fc > 100 → "taquicardia"`

Determinar una bandera alerta (`True`/`False`) que sea `True` si se cumple al menos una de estas condiciones:
- `temperatura >= 38.5`
- `saturacion < 92`
- `clasificacion != "normal"`

**Nota: Recuerda conservar los bloques de código dados, y respetar en tu flujo los nombres de variables en estos.**

```python
# ============================================
# NO MODIFICAR: lectura de datos
# ============================================
fc = int(input("Ingrese los latidos en un minuto: "))
temperatura = float(input("Ingrese a temperatura corporal en °C: "))
saturacion = int(input("Ingrese la saturación de oxígeno (sin el signo %)"))

# ============================================
# COMPLETAR AQUÍ:
# 1. Determina 'clasificacion' ("bradicardia",
# "normal" o "taquicardia") usando if/elif/else.
# 2. Determina 'alerta' (True o False) usando
# una condición compuesta.
# ============================================
if fc < 60:
    clasificacion = "bradicardia"
elif fc <= 100:
    clasificacion = "normal"
else:
    clasificacion = "taquicardia"

if temperatura >= 38.5 or saturacion < 92 or clasificacion != "normal":
  alerta = True
else: alerta = False


# ============================================
# NO MODIFICAR: impresión de resultados
# ============================================
print(f"Frecuencia cardiaca: {clasificacion}")
if alerta:
    print("ALERTA: requiere atencion inmediata")
else:
    print("Sin alerta, paciente estable")
```

### 3. Estructuras cíclicas (20%).
Un fisioterapeuta registra datos de una sesión con un paciente en dos fases:

**Fase 1 — Frecuencia cardíaca:**
Se toman n mediciones de frecuencia cardíaca durante la sesión. Debes contar cuántas de esas mediciones superaron un umbral dado.

**Fase 2 — Duración de ejercicios:**
El paciente reporta la duración (en minutos) de cada ejercicio que realizó, uno por uno. No se sabe cuántos ejercicios hará. El paciente indica que terminó ingresando el valor `0`. Debes acumular el tiempo total y calcular el promedio por ejrecicio.

**Nota: Recuerda conservar los bloques de código dados, y respetar en tu flujo los nombres de variables en estos.**

```python
# ============================================
# NO MODIFICAR: lectura de datos - fase 1
# ============================================
n = int(input("Ingrese la cantidad de mediciones: "))
frecuencias = []
for i in range(n):
    frecuencias.append(int(input("Ingrese la frecuencia: ")))
umbral = int(input("Ingrese el umbral: "))

# ============================================
# CONTADOR
# Cuenta cuántos valores de 'frecuencias' son mayores que 'umbral'
# Guarda el resultado en la variable 'contador'
# ============================================
contador = 0
for f in frecuencias:
  if f > umbral:
    contador += 1

# ============================================
# ACUMULADOR
# Lee duraciones (float) una por una con input(), hasta que se ingrese 0 (no se cuenta)
# Acumula en 'suma_minutos' y cuenta duraciones válidas en 'num_ejercicios'
# Si 'num_ejercicios' es 0, 'promedio' debe quedar en 0.0 (evita división por cero)
# ============================================
suma_minutos = 0.0
num_ejercicios = 0
duracion = float(input("ingrese la duración del ejercicio: "))
while duracion != 0:
  num_ejercicios +=1
  suma_minutos += duracion
  duracion = float(input("ingrese la duración del ejercicio: "))
if num_ejercicios > 0:
  promedio = suma_minutos/num_ejercicios
else:
  promedio = 0.0
# ============================================
# NO MODIFICAR: impresión de resultados
# ============================================
print(f"Mediciones sobre el umbral: {contador}")
print(f"Tiempo total de ejercicio: {suma_minutos:.2f}" " minutos")
print(f"Promedio por ejercicio: {promedio:.2f}" " minutos")
```

Para cada paciente el programa debe:

Al terminar de leer todos los pacientes (al llegar el fin del turno), el programa debe imprimir:

- El total de pacientes atendidos.

#### 4.2 Código (20%).


Ahora si, escribe la solución en código para el problema:

```python
pacientes = []
p_alerta = []
p_bradicardia = []
p_normal = []
p_taquicardia = []
contador = 1
contador_alerta = 0
c_brad = 0
c_norm = 0
c_taq = 0
while True:
  print(f"Paciente {contador}")
  fc = int(input("Ingrese los latidos en un minuto: "))
  if fc != -1:
    temperatura = float(input("Ingrese a temperatura corporal en °C: "))
    saturacion = int(input("Ingrese la saturación de oxígeno (sin el signo %): "))

    if fc < 60:
      clasificacion = "bradicardia"
      c_brad += 1
      p_bradicardia.append(c_brad)
    elif fc <= 100:
      clasificacion = "normal"
      c_norm += 1
      p_normal.append(c_norm)
    else:
      clasificacion = "taquicardia"
      c_taq += 1
      p_taquicardia.append(c_taq)
    if temperatura >= 38.5 or saturacion < 92 or clasificacion != "normal":
      alerta = True
      contador_alerta += 1
      p_alerta.append(contador_alerta)
    else:
      alerta = False

    print(f"Frecuencia cardiaca: {clasificacion}")
    if alerta:
     print("ALERTA: requiere atencion inmediata")
    else:
      print("Sin alerta, paciente estable")

    pacientes.append((fc, temperatura, saturacion, clasificacion, alerta))
    contador += 1

  else:
    break

titulo = "Resultados"
print("=" * 60)
print(titulo.upper().center(60))
print("=" * 60)
print(f"Total de pacientes atendidos: {len(pacientes)}")
print(f"Total de pacientes en alerta: {len(p_alerta)}")
if len(pacientes) == 0:
  print("No hubo pacientes, el promedio es 0")
promedio = (len(p_alerta)/len(pacientes))*100
print(f"El promedio de pacientes en alerta fue {promedio:.2f}%")
print(f"La cantidad de pacientes con bardicaria son: {len(p_bradicardia)}")
print(f"La cantidad de pacientes con fc normal son: {len(p_normal)}")
print(f"La cantidad de pacientes con taquicardia son: {len(p_taquicardia)} ")
```
