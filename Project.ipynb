```python
Cn = int(input("Ingrese la cantidad de notas: "))
contador = 0
Suma = 0
while contador < Cn:
  Nota = float(input(f"Ingrese la nota {contador + 1}: "))
  if Nota < 0 or Nota > 5:

    print("Nota no válida")
  else:

    Suma = Suma + Nota
    contador += 1
Promedio = Suma / Cn
print(f"El promedio de las notas es: {Promedio}")
```
